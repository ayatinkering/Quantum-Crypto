# Why toy Qiskit demos do not scale to RSA-2048 today

This notebook connects small Qiskit demos, such as factoring 15, to the real cryptographic target: RSA-2048.

The short version: Shor's algorithm is asymptotically powerful, but RSA-2048 requires large, fault-tolerant quantum computers. Today's educational circuits are valuable for learning the mechanism, not for breaking production cryptography.

## 1. The toy demo

Factoring 15 is tiny:

- 15 uses only 4 classical bits.
- The useful period for a = 2 is r = 4.
- A laptop simulator can represent the circuit.
- The modular arithmetic can be hard-coded into a small reversible operation.

That is why examples can demonstrate the idea with a handful of qubits.

```python
from math import log2

toy_number = 15
rsa_2048_bits = 2048

print(f"15 needs {toy_number.bit_length()} bits.")
print(f"RSA-2048 modulus needs {rsa_2048_bits} bits.")
print(f"RSA-2048 has about {rsa_2048_bits / toy_number.bit_length():.0f}x more input bits than 15.")
```

## 2. The real scaling issue

The input size is only the start. Shor's algorithm needs reversible modular arithmetic, quantum phase estimation, many controlled operations, and enough error correction to keep the computation alive.

For RSA-2048, the important differences are:

- Thousands of logical qubits may be needed, depending on the implementation.
- The circuit needs a very large number of logical gates.
- Physical hardware is noisy, so each logical qubit must be encoded into many physical qubits.
- Longer circuits require stronger error correction, increasing overhead.
- Current public educational Qiskit examples are not fault-tolerant RSA-breaking systems.

```python
def toy_surface_code_estimate(logical_qubits: int, code_distance: int) -> int:
    """Very rough intuition model: physical ~= logical * 2 * d^2."""
    return logical_qubits * 2 * code_distance**2

logical_qubit_examples = [100, 1000, 5000, 10000]
code_distances = [11, 21, 31]

for logical in logical_qubit_examples:
    print(f"\nLogical qubits: {logical}")
    for distance in code_distances:
        physical = toy_surface_code_estimate(logical, distance)
        print(f"  distance {distance:>2}: ~{physical:,} physical qubits")
```

## 3. Why Qiskit examples still matter

Small Qiskit examples are still useful because they show the mechanism:

- Superposition lets the circuit evaluate many exponent states coherently.
- Modular exponentiation creates periodic structure.
- Quantum Fourier transform extracts period information.
- Classical post-processing turns that period into factors.

The lesson is not "RSA-2048 can be broken today." The lesson is "if large fault-tolerant quantum computers arrive, RSA and ECC need replacement."

## 4. Research takeaway

This is why post-quantum cryptography migration is happening before cryptographically relevant quantum computers exist. Data can be harvested now and decrypted later, and cryptographic migrations take years.

For practical security engineering today, focus on:

- crypto inventory,
- hybrid key exchange,
- ML-KEM for key establishment,
- ML-DSA or SLH-DSA for signatures,
- crypto-agile protocol and certificate management.
